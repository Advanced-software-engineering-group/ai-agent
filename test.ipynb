```python
import anthropic
import os
from dotenv import load_dotenv
```

```python
load_dotenv()
```

```python
def get_claude_response(prompt):
    MODEL = "claude-haiku-5-5"
    anthropic_api_key = os.getenv("API_KEY")
    client = anthropic.Anthropic(api_key=anthropic_api_key)
    response = client.messages.create(
        model=MODEL,
        max_tokens=16000,
        messages=[
            {"role": "user", "content": prompt}
        ]
    )
    return response.content[0].text if response.content and response.content[0].text else ""
```

```python
prompt = "write the numbers from 1 to 20 in a markdown table with 2 columns (number and word). " \
"IMPORTANT: keep only the markdown table and dont write anything else)"
text = get_claude_response(prompt)
with open("file.md", "w", encoding="utf-8") as file:
    file.write(text)
```
